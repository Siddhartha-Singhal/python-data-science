```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('Titanic')
df
```

SURVIVAL COUNT 

```python
sns.countplot(data = df, x = 'survived', hue = 'sex', palette = 'Paired')
plt.xticks([0, 1], ['Not survived', 'Survived'])
plt.ylabel('Count of people')
plt.xlabel(None)
plt.title("Survival count")
plt.show()
```

Survival Rate by Passenger Class

```python
sns.barplot(data = df, x = 'pclass', y = 'survived', hue = 'pclass', palette = 'Pastel2', estimator = np.mean)
plt.title('Survival Rate by Passenger class')
plt.xlabel('Passenger class')
plt.ylabel('Survival rate')
plt.xticks([0, 1, 2], ['1st class', '2nd class', '3rd class'])
plt.show() 
```

Age Distribution by Survival

```python
sns.boxplot(data = df, x = 'survived', y = 'age', hue = 'survived', palette = 'Set2')
plt.xticks([0, 1], ['Not survived', 'Survived'])
plt.title('Age distribution by Survival')
plt.show()
```

Fare Distribution by Class

```python
sns.violinplot(data = df, x = 'pclass', y = 'fare', palette = 'muted')
plt.title('Fare Distribution by Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()
```

Disstribution of Passenger Ages

```python
sns.histplot(df['age'], bins = 30, kde = True, color = 'blue')
plt.title('Distribution of Passenger by their Ages')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

Probability density of Age Distribution for Survivors vs. Non-survivors

```python
sns.kdeplot(data = df, x = 'age', hue = 'survived', fill = True, palette = ['Blue', 'Green'])
plt.title('Age distribution of Survivors vs. Non-surviviors')
plt.xlabel('Age')
plt.ylabel('Density')
plt.legend(['Not survived', 'Survived'])
plt.show()
```

Relationship between fare paid and age, colored by survival status

```python
sns.scatterplot(data = df, x = 'age', y = 'fare', hue = 'survived', palette = 'coolwarm')
plt.title('Fare vs. Age with Survival Status')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

Average fare paid across different passenger classes.

```python
sns.lineplot(data = df, x = 'pclass', y = 'fare', estimator = np.mean, marker = 'o', color = 'purple')
plt.title('Average Fare by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.xticks([1, 2, 3], ['1st class', '2nd class', '3rd class'])
plt.show()
```

Correlation between numerical variables.

```python
corr_matrix = df[['survived', 'pclass', 'age', 'sibsp', 'parch', 'fare']].corr()

sns.heatmap(data = corr_matrix, annot = True, cmap = 'coolwarm', linewidths = 0.5)
plt.title('Correlation heatmap of Titanic features')
plt.show()
```

Pairwise relationships between numerical features.

```python
sns.pairplot(df, vars = ['age', 'fare', 'pclass'], hue = 'survived', palette = 'husl')
plt.show()
```
