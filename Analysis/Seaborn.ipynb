As a data and business analyst, the most essential plots in Seaborn for data exploration and decision-making are:

✔ **Bar Plot** & **Count Plot** → For categorical comparisons

✔ **Box Plot** & **Violin Plot** → For distribution and outliers

✔ **Histogram** & **KDE Plot** → For distribution analysis

✔ **Scatter Plot** & **Line Plot** → For trends and relationships

✔ **Heatmap** & **Pair Plot** → For correlation and multivariate analysis

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('titanic')
df
```

```python
sns.countplot(df, x = 'alive', hue='sex')
plt.title("Survival and non survival count by")
plt.show()
```

```python
sns.countplot(df, y = 'class', hue = 'sex', palette = 'viridis')
plt.title('No of Passenger by class')
plt.show()
```

```python
sns.barplot(df, x = 'class', y = 'survived', hue = 'sex', palette = 'coolwarm')
plt.title('Survival Rate by Class')
plt.show()
```

```python
sns.histplot(df, x = 'age', bins = 30, kde = True, color = 'red')
plt.title('Age distribution')
plt.show()
```

```python
sns.violinplot(df, x='class', y='age', hue='sex', palette='coolwarm')
plt.title('Age distribution by class')
plt.show()
```

```python
df = pd.read_csv("country_wise_latest.csv")
df
```

```python
top_10 = df.nlargest(10, 'Confirmed')
sns.barplot(x = 'Confirmed', y = 'Country/Region', data = top_10, palette = 'viridis')
plt.title("Top 10 countries with max deaths")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.lineplot(x='Confirmed last week', y='1 week change', data=df, marker='o')
plt.title('Weekly Case Growth')
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.scatterplot(x='Confirmed', y='Active', data=df, hue='WHO Region', palette='viridis')
plt.title('Active Case vs Confirmed case')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(df['Deaths / 100 Cases'], bins = 30, kde=True, color='red')
plt.title("Distribution of Death rate")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(x='WHO Region', y='New cases', data=df, palette='pastel')
plt.xticks(rotation=45)
plt.title('New Cases by WHO Region')
plt.show()
```

```python
plt.figure(figsize=(8,8))
df.groupby('WHO Region')['Confirmed'].sum().plot.pie(autopct='%1.1f%%', cmap='coolwarm', startangle=140)
plt.ylabel('')
plt.title('Confirmed Cases Distribution by WHO Region')
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.relplot(data=df, x="Deaths", y="Active",hue="WHO Region",height=6)
plt.title("CASES")
plt.show()
```

```python
sns.pairplot(df[['Confirmed', 'Deaths', 'Recovered', 'Active']], diag_kind='kde', plot_kws={'alpha':0.5})
plt.show()
```
