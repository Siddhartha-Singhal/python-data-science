```python
import pandas as pd
import plotly.express as px
import nbformat
```

```python
df = pd.read_csv("country_wise_latest.csv")
df
```

```python
top_10 = df.nlargest(10, 'Confirmed')
fig = px.bar(top_10, x='Country/Region', y='Confirmed', title='Top 10 countries with highest confirmed cases', text='Confirmed', color='Confirmed', template='plotly_dark')
fig.update_traces(textposition='outside')
fig.show()
```

```python
fig = px.pie(df, names = 'WHO Region', values = 'Confirmed', title = 'Confirmed cases per WHO region', template = 'plotly_dark', color_discrete_sequence = px.colors.sequential.Rainbow, height = 500, width = 800)
fig.show()
```

```python
trendline =  df.sort_values(by='1 week % increase', ascending=False).head(20) # this will sort the data by 1% increase in descending order and select top 20 

fig = px.line(trendline, x = 'Country/Region', y = '1 week % increase', title = 'Weekly % increase in COVID-19 cases', template = 'plotly_dark', markers = True, width=800, height=500)
fig.update_traces(marker=dict(size=8, symbol='circle-open'))
fig.show()
```
