```python
import pandas as pd

# Sample DataFrame
data = {
    'Age': [25, 30, 22, 40],
    'Salary': [50000, 60000, 45000, 80000],
    'Department': ['HR', 'IT', 'Finance', 'IT']
}

df = pd.DataFrame(data)

# Default describe (Only numeric columns)
print(df.describe())
```

```python
df.describe(include='object')  # Only categorical columns
df.describe(include='all')  # Both numerical and categorical columns
```
